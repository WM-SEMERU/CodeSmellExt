### Imports

```python
import scipy.stats as stats
import pandas as pd
```

### Analysis

```python
def expand_likert_counts(likert_counts):
    """Expands a list of Likert-scale counts into a flat list of individual ratings."""
    responses = []
    for i, count in enumerate(likert_counts):
        responses.extend([i + 1] * count)  # Likert scale starts at 1
    return responses

def analyze_expanded_survey(control_answers, treatment_answers):
    results = []

    for snippet in control_answers:
        for question in ['Q1', 'Q2', 'Q3']:
            control_raw = expand_likert_counts(control_answers[snippet][question])
            treatment_raw = expand_likert_counts(treatment_answers[snippet][question])
            
            if control_raw and treatment_raw:  # ensure both groups have data
                u_stat, p_val = stats.mannwhitneyu(control_raw, treatment_raw, alternative='two-sided')
                significance = "✅ Significant" if p_val < 0.05 else "❌ Not Significant"
                results.append({
                    "Snippet": snippet,
                    "Question": question,
                    "U Statistic": u_stat,
                    "p-value": round(p_val, 4),
                    "Significance": significance
                })

    return pd.DataFrame(results)
```

```python
df_results = analyze_expanded_survey(params['control_answers'], params['treatment_answers'])
print(df_results)
```
