# Dataset preparation for code-completion

```python
def default_params(): 
    return {
        'dataset' : '/workspaces/CodeSmells/semeru-datasets/code_smells/curated_500.json',
        'output_path' : '/workspaces/CodeSmells/semeru-datasets/code_smells/generation', 
        'tokenizer' : 'codellama/CodeLlama-7b-hf',
        'cache_dir': '/workspaces/CodeSmells/datax/hugging_face_cache',
        'sampling_size' : 500,
        'prompt_text' : """complete the following incomplete Python function:\n"""
    }
params = default_params()
```

### Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set_theme()
from collections import Counter
import plotly.express as px
from transformers import AutoTokenizer
```

### Load dataset

```python
completion_df = pd.read_json(params['dataset'])
```

### Generate prompts

```python
def remove_percentage_of_string(text, percentage=50):
    text_length = len(text)
    chars_to_remove = int((percentage / 100) * text_length)
    chop_index = text_length - chars_to_remove
    return text[:chop_index]
generate_prompt = lambda code: f"{params['prompt_text']}{remove_percentage_of_string(code)}"
```

```python
completion_df['prompt'] = completion_df['code'].map(lambda code: generate_prompt(code))
```

### Store dataset 

```python
completion_df.to_json(f"{params['output_path']}/curated_{params['sampling_size']}.json", index=False)
```
