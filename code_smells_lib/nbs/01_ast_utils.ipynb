```python
#| export
from tree_sitter import Language, Parser
import tree_sitter_python as tspython
```

```python
#| export
def extract_python_elements(source_code: str):
    """
    Extract identifiers, comments, and strings from Python source code using Tree-sitter.
    Returns a dictionary with keys: 'identifiers', 'comments', 'strings'.
    Each value is a list of Tree-sitter nodes.
    """
    language = Language(tspython.language())
    parser = Parser(language)

    tree = parser.parse(source_code.encode("utf8"))
    root_node = tree.root_node

    result = {
        'identifiers': [],
        'comments': [],
        'strings': []
    }

    def traverse(node):
        if node.type == 'identifier':
            result['identifiers'].append(node)
        elif node.type == 'comment':
            result['comments'].append(node)
        elif node.type == 'string':
            result['strings'].append(node)

        for child in node.children:
            traverse(child)

    traverse(root_node)
    return result
```

### Examples

```python
#| hide
complex_code = """
# Module-level comment

class Greeter:
    def __init__(self, greeting):
        self.greeting = greeting  # Instance variable

    def greet(self, name):
        '''Greet someone by name.'''
        print(f"{self.greeting}, {name}!")  # Print greeting

def farewell(name):
    \"\"\"Say goodbye.\"\"\"
    print("Goodbye, " + name)
"""

elements = extract_python_elements(complex_code)

print("Identifiers:", [node.text.decode() for node in elements['identifiers']])
print("Comments:", [node.text.decode() for node in elements['comments']])
print("Strings:", [node.text.decode() for node in elements['strings']])
```
