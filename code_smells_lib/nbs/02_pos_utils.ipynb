```python
#| export
import spacy
from collections import Counter
import subprocess
import importlib.util
```

```python
#| export
UNIVERSAL_SEMANTIC_TAGS = ['NOUN', 'PROPN', 'VERB', 'ADJ', 'ADV', 'NUM', 'INTJ']
```

```python
#| export
def load_spacy_model(model_name="en_core_web_sm"):
    try:
        return spacy.load(model_name)
    except OSError:
        print(f"Downloading spaCy model '{model_name}'...")
        subprocess.run(["python", "-m", "spacy", "download", model_name], check=True)
        return spacy.load(model_name)
```

```python
#| export
def count_pos_tags(text: str) -> dict:
    nlp = load_spacy_model()
    doc = nlp(text)
    pos_counts = Counter(token.pos_ for token in doc)
    return dict(pos_counts)
```

```python
#| export
def get_all_pos_types(model_name="en_core_web_sm") -> list:
    try:
        nlp = spacy.load(model_name)
    except OSError:
        import subprocess
        subprocess.run(["python", "-m", "spacy", "download", model_name], check=True)
        nlp = spacy.load(model_name)

    return list(nlp.get_pipe("tagger").labels)
```

### Example

```python
#| hide
# Example usage of count_pos_tags
sample_text = "The quick brown fox jumps over the lazy dog."
pos_counts = count_pos_tags(sample_text)
print(pos_counts)
get_all_pos_types()
```
